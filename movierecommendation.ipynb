```python
import ast

def fetch_director(text):
    directors = []
    for i in ast.literal_eval(text):
        if i.get('job') == 'Director':
            directors.append(i.get('name'))
    return directors

# Create a new column called 'directors'
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vector = cv.fit_transform(new_df['tags']).toarray()
vector.shape
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()

def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)  # Return after the loop ends

# Apply the stem function to the 'tags' column
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
new_df['tags'][0]
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity.shape
```

```python
def recommend(movie):
    # Step 1: Find the index of the movie title
    movie_index = new_df[new_df['title'] == movie].index[0]
    
    # Step 2: Get similarity scores for this movie
    distances = similarity[movie_index]
    
    # Step 3: Sort the distances and get the top 5 most similar movies (excluding the input movie itself)
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    # Step 4: Print recommended movie titles
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend("Thor")
```
